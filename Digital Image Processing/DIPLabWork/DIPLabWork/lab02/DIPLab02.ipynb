```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
for i in range(rows):
  for j in range(cols):
    negative_image[i,j] = 255 - image[i,j]
```

```python
fig = plt.figure(figsize=(10,5))
ax1 = fig.add_subplot(1,2,1)
ax1.imshow(image,cmap='gray')
ax2 = fig.add_subplot(1,2,2)
ax2.imshow(negative_image,cmap='gray')
```

# Log Transformation

```python
image = cv2.imread('Lena-original.jpg', cv2.IMREAD_GRAYSCALE).astype(np.float32)
c = 255 / np.log(1+np.max(image))

log_img = np.zeros_like(image)

for i in range(rows):
  for j in range(cols):
    log_img[i,j] = c * np.log(1 + image[i, j])

```

```python
fig = plt.figure(figsize=(10,5))
ax1 = fig.add_subplot(1,2,1)
ax1.set_title('Original Image')
ax1.imshow(image,cmap='gray')
ax2 = fig.add_subplot(1,2,2)
ax2.set_title('Log Transformation')
ax2.imshow(log_img,cmap='gray')
```

```python
print(log_img)
print(image)
```

# Gamma correction

```python
gamma = 2.2
c = 1
# Create output image
gamma_img = np.zeros_like(image, dtype=np.uint8)

# Apply gamma correction using nested loop
for i in range(rows):
    for j in range(cols):
        gamma_img[i, j] = c * (image[i,j] ** gamma)


```

```python
print(gamma_img)
print(image)
```

```python
# Show images
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(image, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Gamma Corrected")
plt.imshow(gamma_img, cmap='gray')
plt.tight_layout()
plt.show()
```

# Thresholding

```python
img_th = np.zeros_like(image)
for i in range(rows):
  for j in range(cols):
    if(image[i,j] < 128):
      img_th[i,j] = 0
    else:
      img_th[i,j] = 255
```

```python
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(image, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Thresholding")
plt.imshow(img_th, cmap='gray')
plt.tight_layout()
plt.show()
```

```python
print(img_th)
print(image)
```

# Constrast Stretching

```python
contrast_img = np.zeros_like(image)
for i in range(rows):
  for j in range(cols):
    contrast_img[i,j] = (255 * (image[i,j] - np.min(image))) / (np.max(image) - np.min(image))
```

```python
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(image, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Contrast Stretching")
plt.imshow(contrast_img, cmap='gray')
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image as float32 for arithmetic safety
image = cv2.imread('Lena-original.jpg', cv2.IMREAD_GRAYSCALE).astype(np.float32)
rows, cols = image.shape

# ---------------- Negative ----------------
negative_image = np.zeros_like(image, dtype=np.float32)
for i in range(rows):
    for j in range(cols):
        negative_image[i, j] = 255 - image[i, j]

# ---------------- Log Transform ----------------
c = 255 / np.log(1 + np.max(image))
log_img = np.zeros_like(image, dtype=np.float32)
for i in range(rows):
    for j in range(cols):
        log_img[i, j] = c * np.log(1 + image[i, j])

# ---------------- Gamma Correction (γ = 2.2) ----------------
gamma = 2.2
gamma_img = np.zeros_like(image, dtype=np.float32)
for i in range(rows):
    for j in range(cols):
        norm_val = image[i, j] / 255.0          # 1. normalise
        gamma_img[i, j] = 255 * (norm_val ** gamma)  # 2+3

# ---------------- Binary Thresholding ----------------
img_th = np.zeros_like(image, dtype=np.float32)
for i in range(rows):
    for j in range(cols):
        img_th[i, j] = 0 if image[i, j] < 128 else 255

# ---------------- Contrast Stretching ----------------
contrast_img = np.zeros_like(image, dtype=np.float32)
imin, imax = np.min(image), np.max(image)
for i in range(rows):
    for j in range(cols):
        contrast_img[i, j] = 255 * (image[i, j] - imin) / (imax - imin)

# Cast to uint8 for display
display_imgs = [image, negative_image, log_img, gamma_img, img_th, contrast_img]
display_imgs = [np.uint8(np.clip(im, 0, 255)) for im in display_imgs]

# Display results
titles = ['Original', 'Negative', 'Log', 'Gamma', 'Threshold', 'Contrast Stretch']
plt.figure(figsize=(15, 7))
for k, (title, im) in enumerate(zip(titles, display_imgs), 1):
    plt.subplot(2, 3, k)
    plt.imshow(im, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.tight_layout()
plt.show()
```
