```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
```

#Difference between sine wave and cos wave

```python
srate=256
t=np.arange(0.,1.,1/srate)


sinewave=3*np.sin(2*np.pi*5*t)
coswave=3*np.cos(2*np.pi*5*t)

plt.figure(figsize=(12,6))
style.use('dark_background')

plt.plot(t,sinewave,'y',linewidth=3,label='sine wave',color='red')
plt.plot(t,coswave,'y',linewidth=3,label='cos wave',color='blue')
plt.xlabel('Time'),plt.ylabel('Amplitude')
plt.title('Sine and Cosine wave comparison',fontsize=20)
plt.legend()
plt.xlim([0,1])
plt.show()
```

```python
srate=256 #sampling rate in Hz
t=np.arange(0.,1.,1/srate)#time vector in seconds

x1=np.sin(2*np.pi*2*t) #sine wave

x2=2*np.cos(2*np.pi*4*t) #cosine wave

x3=x1+x2 #sum of the two waves

DC=2 #DC shift
x4=x2+x1+DC #DC shift added

```

```python
#ploting
plt.figure(figsize=(24,14))
plt.suptitle('constructing a wave with sine, cosine and DC',fontsize=40)
style.use('dark_background')

plt.subplot(2,2,1)
plt.plot(t,x1,'r',linewidth=3)
plt.title('sine wave of amplitude 1 and frequency 2Hz ',fontsize=20)
plt.xlabel('Time in sec',fontsize=20),plt.ylabel('Amplitude',fontsize=20)
# plt.xlim([0,1])

plt.subplot(2,2,2)
plt.plot(t,x2,'b',linewidth=3)
plt.title('cosine wave of amplitude 2 and frequency 4Hz',fontsize=20)
plt.xlabel('Time in sec'),plt.ylabel('Amplitude')
# plt.xlim([0,1])

plt.subplot(2,2,3)
plt.plot(t,x3,'g',linewidth=3)
plt.title('sum of sine and cosine',fontsize=20)
plt.xlabel('Time'),plt.ylabel('Amplitude')
# plt.xlim([0,1])

plt.subplot(2,2,4)
plt.plot(t,x4,'y',linewidth=3)
plt.title('sum of sine and cosine with DC shift',fontsize=20)
plt.xlabel('Time'),plt.ylabel('Amplitude')
# plt.xlim([0,1])

plt.show()
```

```python
srate=256
nyquist=0.5*srate
t=np.arange(0.,1.,1/srate)

x1=5*np.sin(2*np.pi*2*t) #sine wave
x2=2*np.sin(2*np.pi*4*t) #sine wave
x3=7*np.sin(2*np.pi*6*t) #sine wave

x=x1+x2+x3 #sum of the three waves

Ck=np.zeros(len(t),dtype=complex)

for freq in range(0,len(t)):
    #create complex sine wave for each frequency and multiply with the signal to get the fourier transform at that frequency
    csw=np.exp(-1j*2*np.pi*freq*t)
    Ck[freq]=np.sum(np.multiply(x,csw))
    
amps=2*np.abs(Ck)/len(t)

# nyquist=0.5*srate
Hz=np.linspace(0,nyquist,math.floor(len(t)/2)+1)


plt.figure(figsize=(24,14))
plt.suptitle('Fourier Transform of a signal',fontsize=40)
style.use('dark_background')
plt.rcParams['xtick.labelsize']=15
plt.rcParams['ytick.labelsize']=15

plt.subplot(2,2,1)
plt.plot(t,x,linewidth=3)
plt.title('sum of sine waves of 2,4 and 6 Hz',fontsize=20)
plt.xlabel('Time in sec',fontsize=20),plt.ylabel('Amplitude',fontsize=20)

plt.subplot(2,2,2)
markerline,stemlines,baseline=plt.stem(Hz,amps[range(0,len(Hz))])
plt.setp(stemlines,'linewidth',5)
# plt.plot(Hz,amps[:len(Hz)],'b',linewidth=3)
plt.xlim([0,10])
plt.ylim([0,8])
plt.title('Fourier Transform of the signal',fontsize=20)
plt.xlabel('Frequency in Hz',fontsize=20),plt.ylabel('Amplitude',fontsize=20)
plt.ylabel('Amplitude',fontsize=20)

plt.show()



```
