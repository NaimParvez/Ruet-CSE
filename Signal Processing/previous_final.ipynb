```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
```

```python
def transform_signal(n_vals, x_vals):
    # Interpolation function for any custom n_vals
    interp_func = interp1d(n_vals, x_vals, kind='linear', fill_value=0, bounds_error=False)

    # New n range for transformed output
    n_new = np.arange(-5, 5)  # You can adjust this as needed
    n_transformed = -2 * n_new + 2
    y_vals = 2 * interp_func(n_transformed)

    # Print transformed values
    print("n_new:", n_new)
    print("Transformed y[n]:", y_vals)

    # Plotting
    plt.figure(figsize=(10, 5))
    plt.stem(n_vals, x_vals, linefmt='b-', markerfmt='bo', basefmt='k-', label='Original x[n]')
    plt.stem(n_new, y_vals, linefmt='r--', markerfmt='ro', basefmt='k-', label='Transformed y[n] = 2x[-2n + 2]')
    plt.legend()
    plt.xlabel('n')
    plt.ylabel('Amplitude')
    plt.title('Signal Transformation with Custom Indices')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
# Example usage with custom indices
n_vals = np.array([-2, -1, 0, 1, 2, 3])
x_vals = np.array([2, 5, -9, 1, 0, 3])
transform_signal(n_vals, x_vals)
```
