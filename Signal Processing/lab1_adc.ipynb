```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
# Generate the analog signal
t = np.linspace(0, 20, 1000)  # Time vector (0 to 20 seconds, 1000 points)
x = 0.85 ** t  # Analog signal: x(t) = (0.85
```

```python
# Sampling (reduce the resolution in time)
sampling_rate = 100  # 50 samples per second
t_sampled = np.linspace(0, 20, sampling_rate)
x_sampled = 0.85 ** t_sampled  # Sampled values of the signal
```

```python

# Quantization (reduce the resolution in amplitude)
quantization_levels = 512  # Number of levels (3-bit resolution)
x_quantized = np.round(x_sampled * (quantization_levels - 1))  # Normalize and quantize
x_quantized = x_quantized / (quantization_levels - 1)  # Map back to 0 to 1 range
```

```python
plt.figure(figsize=(15, 10))
plt.suptitle("Analog to Digital Conversion (ADC)", fontsize=20)
style.use('dark_background')
```

```python
# Analog Signal Plot
plt.subplot(2, 1, 1)
plt.plot(t, x, 'r', label='Analog Signal x(t) = (0.85)^t', linewidth=2)
plt.xlabel("Time (s)", fontsize=15)
plt.ylabel("Amplitude", fontsize=15)
plt.legend(fontsize=12)
plt.grid(True)
```

```python
# Digital Signal Plot (Sampling + Quantization)
plt.subplot(2, 1, 2)
plt.plot(t, x, 'r', label='Original Analog Signal', alpha=0.5, linestyle='--')
plt.stem(t_sampled, x_quantized, linefmt='g-', markerfmt='go', basefmt='k-', label='Digital Signal (Sampled & Quantized)')
plt.xlabel("Time (s)", fontsize=15)
plt.ylabel("Amplitude", fontsize=15)
plt.legend(fontsize=12)
plt.grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
