```python
column= focus.to_numpy().sum(axis=0)
column
focus['trestbps'][0]
```

```python
focus=focus.to_numpy()
```

```python
total_sum=np.sum(column)
total_sum
```

```python
chi_score=0
total_sum=np.sum(column)
for i in range(focus.shape[0]):
    row_sum=focus['trestbps'][i][0]+focus[i][1]
    ex1=(row_sum*column[0])/total_sum
    ex2=(row_sum*column[1])/total_sum
    chi_score+=(((focus[i][0]-ex1)**2)/focus[i][0])+(((focus[i][1]-ex2)**2)/focus[i][1])
chi_score
    
```
