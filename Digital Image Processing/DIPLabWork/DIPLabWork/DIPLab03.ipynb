```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('shohag.jpg', cv2.IMREAD_GRAYSCALE)
image.shape
```

```python
image_1D = image.flatten()
```

```python
histogram = np.zeros(256,dtype=float)
#occurance
for i in image_1D:
  histogram[i]+=1

print(histogram)
```

```python
# sum = 0
# for i in range(256):
#   sum+=histogram[i]

# print(sum)
# print(320*320)
sum = np.sum(histogram)
```

```python
pdf = histogram.astype(float) / sum
```

```python
cumsum = 0
for i in range(256):
  cumsum+=255*pdf[i]
  pdf[i]=cumsum

print(pdf)
```

```python
normalized = np.round(pdf).astype(int)
```

```python
equalized_image = np.zeros_like(image, dtype=int)

for i in range(len(image)):
    equalized_image[i] = normalized[image[i]]
```

```python
equalized_image = equalized_image.reshape((image.shape[0], image.shape[1]))

# Display the original and equalized images
plt.figure(figsize=(12, 10))

# Original Image
plt.subplot(2, 2, 1)
plt.imshow(image.reshape((image.shape[0], image.shape[1])), cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Equalized Image
plt.subplot(2, 2, 2)
plt.imshow(equalized_image, cmap='gray')
plt.title('Equalized Image')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.hist(image_1D,bins=256,range=(0, 255))
plt.title('Histogram of Original Image')


plt.subplot(2, 2, 4)
plt.hist(equalized_image.flatten(),bins=256,range=(0,255))
plt.title('Histogram of Equalized Image')
plt.tight_layout()

plt.show()
```
