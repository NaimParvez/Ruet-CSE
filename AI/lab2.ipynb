```python
# Creating a tuple
my_tuple = (1, 2, 3, 4, 5)
# to create a tuple with a single element, we need to add a comma after the element
my_tuple1=(1,)

# Accessing elements
first_element = my_tuple[0]
last_element = my_tuple[-1]

# Slicing a tuple
sub_tuple = my_tuple[1:4]

# Unpacking a tuple
a, b, c, d, e = my_tuple

# Tuple length
tuple_length = len(my_tuple)

# Printing the results
print("Tuple:", my_tuple)
print("First element:", first_element)
print("Last element:", last_element)
print("Sub tuple:", sub_tuple)
print("Unpacked values:", a, b, c, d, e)
print("Tuple length:", tuple_length)


#since tuple is immutable, we cannot change the values of a tuple. to make changes to a tuple, we need to convert it to a list
list3=list(my_tuple)
list3
type(list3)
```

```python
# Creating a set
my_set = {1, 2, 3, 4, 5}
my_set1 = {1, 2, 3, 4, 5, 5, 5, 5, 5} #set does not allow duplicate values

# Adding an element to the set
my_set.add(6)

# Removing an element from the set
my_set.remove(3)

# Checking if an element is in the set
is_in_set = 4 in my_set

# Getting the length of the set
set_length = len(my_set)

# Printing the results
print("Set:", my_set)
print("Is 4 in the set?", is_in_set)
print("Set length:", set_length)

#in set we cannot access elements using index, since set is unordered, we cannot access elements using index
# we can convert a set to a list to access elements using index
list4=list(my_set)
list4[0]

my_set.difference(my_set1) #returns the elements that are in my_set but not in my_set1

my_set.issubset(my_set1) #returns True if my_set is a subset of my_set1
```

```python
# Creating a dictionary
my_dict = {
    'Name': 'Alice',
    'Age': 25,
    'Gender': 'Female',
    'Occupation': 'Engineer'
}

# Creating an empty dictionary
empty_dict = {}
empty_dict1 = dict()

# Accessing values
name = my_dict['Name']
age = my_dict['Age']

# Adding a new key-value pair
my_dict['City'] = 'New York'

# Updating an existing value
my_dict['Age'] = 26

# Removing a key-value pair
del my_dict['Occupation']

# Checking if a key exists
has_gender = 'Gender' in my_dict

# Getting the length of the dictionary
dict_length = len(my_dict)

# Printing the results
print("Dictionary:", my_dict)
print("Name:", name)
print("Age:", age)
print("Has Gender key?", has_gender)
print("Dictionary length:", dict_length)


for i in my_dict:
    print(i,my_dict[i])
```

```python
# Base class
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclass must implement abstract method")

# Derived class
class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"

# Derived class
class Cat(Animal):
    def speak(self):
        return f"{self.name} says Meow!"

# Example usage
dog = Dog("Buddy")
cat = Cat("Whiskers")

print(dog.speak())  # Output: Buddy says Woof!
print(cat.speak())  # Output: Whiskers says Meow!
```
