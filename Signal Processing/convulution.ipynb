WEZNEY METHOD

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
srate = 256 # Hz
t  = np.arange(0,3,1/srate)
pnts = len(t)

print(pnts)
print(t)
```

```python
# creating a noiseless signal
x = np.sin(2*np.pi * 2 * t)

# creating a random noise
noise  = 5 * np.random.randn(pnts)
# Adding noise on signal
Noisysignal = x + noise

x=Noisysignal
```

```python
filter = 2*np.ones(50)/10  # Moving Average Filter
h=filter
```

```python
# x=np.array([1,2,3,1])
# h=np.array([1,2,1,-1])
n = len(x) + len(h) - 1
y = np.zeros(n)
for i in range(n):
    for j in range(len(x)):
        if i - j >= 0 and i - j < len(h):
            y[i] += x[j] * h[i - j]
print(y)
```

```python
plt.figure(figsize = (20,12))
style.use('dark_background')
plt.rcParams['xtick.labelsize'] = 25
plt.rcParams['ytick.labelsize'] = 25
plt.plot(t,Noisysignal, label = 'Noisy Signal')
plt.legend(fontsize = 20)
plt.show()
```
